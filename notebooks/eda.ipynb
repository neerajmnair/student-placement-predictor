```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style()
```

```python
df = pd.read_csv("../data/raw/placement.csv")
df.head()
```

## How many students got placed?

```python
plt.figure(figsize=(6,4))
sns.countplot(x="PlacementStatus",data=df) #builds a bar chart
plt.title("Placement Status Distribution")
plt.xlabel("Placement Status")
plt.ylabel("No of students")

plt.show()
```

## Observation
This dataset contains more students who are not placed compared to those who are placed.

## Does CGPA influence placement?

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="PlacementStatus", y="CGPA", data=df) #boxplot let us compare the range of CG in both categories
plt.title("CGPA vs Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("CGPA")

plt.show()
```

## Observation
Students with higher CGPA generally showed better placement outcomes.

## Does Internships affect Placement?

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Internships", hue="PlacementStatus", data=df) 
plt.title("Internships vs Placement Status")
plt.xlabel("No of Internships")
plt.ylabel("Number of Students")

plt.show()
```

### Observation

Students with a higher number of internships generally exhibited better placement outcomes. This suggests practical experience positively influences employability.

## Does placement trainig impact Placement Status?

```python
plt.figure(figsize=(6,4))
sns.countplot(x="PlacementTraining", hue="PlacementStatus", data=df) 
plt.title("Placement Training vs Placement Status")
plt.xlabel("Placement Training")
plt.ylabel("Number of Students")

plt.show()
```

## Observation

As predictable, students who received placent Training received better placement outcomes.

## Does Aptitude Test Score affect Placement Status

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="PlacementStatus", y="AptitudeTestScore", data=df) 
plt.title("Aptitude Test Score vs Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("Aptitude Test Score")

plt.show()
```

## Observation

Students who are placed have higher Aptitude test scores.

## Do Soft Skills Influence Placement?

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="PlacementStatus", y="SoftSkillsRating", data=df) 
plt.title("Soft Skills Rating vs Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("Soft Skills Rating (on 5)")

plt.show()
```

## Do Extra-Curricular Activities affect Placement?

```python
plt.figure(figsize=(6,4))
sns.countplot(x="ExtracurricularActivities",hue="PlacementStatus",  data=df) 
plt.title("Extra-curricular Activities vs Placement Status")
plt.xlabel("Extra-curricular Activities")
plt.ylabel("No of Students")

plt.show()
```

## Observation
It is apparent from the graph that most of the people who are placed were involved in extracurricular activities

## Which features are most correlated with placement?

This analysis identifies relationships among numerical variables and helps determine which features most strongly influence placement outcomes.

```python
df_heatmap = df.copy()
```

```python
df_heatmap["PlacementStatus"]=(
    df_heatmap["PlacementStatus"].map({
            "NotPlaced":0,
            "Placed":1
    }))
```

```python
numeric_df=df_heatmap.select_dtypes(
    include=["int64","float64"]
)
```

```python
plt.figure(figsize=(8,4))

sns.heatmap(
    numeric_df.corr(),
    cmap="coolwarm",
    annot=True
)
plt.title("Correlation Heatmap")
plt.show()
```

### Observation

CGPA, aptitude test score, and soft skills rating demonstrated relatively stronger positive correlations with placement outcomes. This suggests academic performance and employability skills are important factors influencing placements.
